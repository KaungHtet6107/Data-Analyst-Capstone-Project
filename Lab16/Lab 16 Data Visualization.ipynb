**Read the CSV File into a Pandas DataFrame**

**Create a SQLite Database and Insert the Data**

Now, let's create a new SQLite database (`survey-data.sqlite`) and insert the data from the DataFrame into a table using the sqlite3 library:


```python
import sqlite3

# Create a connection to the SQLite database
conn = sqlite3.connect('survey-data.sqlite')

# Write the dataframe to the SQLite database
df.to_sql('main', conn, if_exists='replace', index=False)


# Close the connection
conn.close()
```

Verify that the data has been correctly inserted into the SQLite database by running a simple query:


```python
# Reconnect to the SQLite database
conn = sqlite3.connect('survey-data.sqlite')

# Run a simple query to check the data
QUERY = "SELECT * FROM main LIMIT 5"
df_check = pd.read_sql_query(QUERY, conn)

# Display the results
print(df_check)
```

## Hands-on Lab


### Visualizing the Distribution of Data

**Histograms**

Plot a histogram of CompTotal (Total Compensation).


```python
## Write your code here
# Get CompTotal data from the database
QUERY = """
SELECT CompTotal
FROM main
WHERE CompTotal IS NOT NULL
"""

comp_data = pd.read_sql_query(QUERY, conn)

# Convert to numeric
comp_data["CompTotal"] = pd.to_numeric(comp_data["CompTotal"], errors="coerce")
comp_data = comp_data.dropna()

# Create histogram
plt.figure(figsize=(10, 6))
plt.hist(comp_data["CompTotal"], bins=50)

plt.title("Distribution of Total Compensation")
plt.xlabel("Total Compensation")
plt.ylabel("Number of Respondents")
plt.show()
```

**Box Plots**

Plot a box plot of Age.


```python
## Write your code here
# Get Age data from the database
QUERY = """
SELECT Age
FROM main
WHERE Age IS NOT NULL
"""

age_data = pd.read_sql_query(QUERY, conn)

# Convert age groups to approximate numeric values
age_mapping = {
    "Under 18 years old": 17,
    "18-24 years old": 21,
    "25-34 years old": 29,
    "35-44 years old": 39,
    "45-54 years old": 49,
    "55-64 years old": 59,
    "65 years or older": 70
}

age_data["Age_numeric"] = age_data["Age"].map(age_mapping)

# Create box plot
plt.figure(figsize=(8, 6))
plt.boxplot(age_data["Age_numeric"].dropna())

plt.title("Box Plot of Age")
plt.ylabel("Age")
plt.show()
```

### Visualizing Relationships in Data

**Scatter Plots**

Create a scatter plot of Age and WorkExp.


```python
## Write your code here
# Get Age and WorkExp data
QUERY = """
SELECT Age, WorkExp
FROM main
WHERE Age IS NOT NULL
AND WorkExp IS NOT NULL
"""

scatter_data = pd.read_sql_query(QUERY, conn)

# Convert Age groups to numeric values
scatter_data["Age_numeric"] = scatter_data["Age"].map(age_mapping)

# Convert WorkExp to numeric
scatter_data["WorkExp"] = pd.to_numeric(
    scatter_data["WorkExp"], 
    errors="coerce"
)

scatter_data = scatter_data.dropna(
    subset=["Age_numeric", "WorkExp"]
)

# Create scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(
    scatter_data["Age_numeric"],
    scatter_data["WorkExp"],
    alpha=0.5
)

plt.title("Age vs Work Experience")
plt.xlabel("Age")
plt.ylabel("Work Experience")
plt.show()
```

**Bubble Plots**

Create a bubble plot of `TimeSearching` and `Frustration` using the Age column as the bubble size.


```python
## Write your code here
# Get required columns
QUERY = """
SELECT TimeSearching, Frustration, Age
FROM main
WHERE TimeSearching IS NOT NULL
AND Frustration IS NOT NULL
AND Age IS NOT NULL
"""

bubble_data = pd.read_sql_query(QUERY, conn)

# Convert Age groups to numeric
bubble_data["Age_numeric"] = bubble_data["Age"].map(age_mapping)

# Convert TimeSearching and Frustration to numeric
bubble_data["TimeSearching"] = pd.to_numeric(
    bubble_data["TimeSearching"],
    errors="coerce"
)

bubble_data["Frustration"] = pd.to_numeric(
    bubble_data["Frustration"],
    errors="coerce"
)

bubble_data = bubble_data.dropna(
    subset=["TimeSearching", "Frustration", "Age_numeric"]
)

# Create bubble plot
plt.figure(figsize=(10, 6))

plt.scatter(
    bubble_data["TimeSearching"],
    bubble_data["Frustration"],
    s=bubble_data["Age_numeric"] * 5,
    alpha=0.4
)

plt.title("Time Searching vs Frustration")
plt.xlabel("Time Searching")
plt.ylabel("Frustration")
plt.show()
```

### Visualizing Composition of Data

**Pie Charts**

Create a pie chart of the top 5 databases(`DatabaseWantToWorkWith`) that respondents wish to learn next year.


```python
## Write your code here
# Get database preferences
QUERY = """
SELECT DatabaseWantToWorkWith
FROM main
WHERE DatabaseWantToWorkWith IS NOT NULL
"""

database_data = pd.read_sql_query(QUERY, conn)

# Split multiple answers and count each database
database_counts = (
    database_data["DatabaseWantToWorkWith"]
    .str.split(";")
    .explode()
    .str.strip()
    .value_counts()
)

# Get top 5
top_5_databases = database_counts.head(5)

# Create pie chart
plt.figure(figsize=(8, 8))

plt.pie(
    top_5_databases.values,
    labels=top_5_databases.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Top 5 Databases Respondents Want to Work With")
plt.show()
```

**Stacked Charts** 

Create a stacked bar chart of median `TimeSearching` and `TimeAnswering` for the age group 30 to 35.


```python
## Write your code here
# Get data for the 25-34 age group
QUERY = """
SELECT TimeSearching, TimeAnswering
FROM main
WHERE Age = '25-34 years old'
AND TimeSearching IS NOT NULL
AND TimeAnswering IS NOT NULL
"""

time_data = pd.read_sql_query(QUERY, conn)

# Convert to numeric
time_data["TimeSearching"] = pd.to_numeric(
    time_data["TimeSearching"],
    errors="coerce"
)

time_data["TimeAnswering"] = pd.to_numeric(
    time_data["TimeAnswering"],
    errors="coerce"
)

time_data = time_data.dropna()

# Calculate medians
median_time_searching = time_data["TimeSearching"].median()
median_time_answering = time_data["TimeAnswering"].median()

# Create stacked bar chart
plt.figure(figsize=(8, 6))

plt.bar(
    ["Age 30-35"],
    [median_time_searching],
    label="TimeSearching"
)

plt.bar(
    ["Age 30-35"],
    [median_time_answering],
    bottom=[median_time_searching],
    label="TimeAnswering"
)

plt.title("Median Time Searching and Time Answering")
plt.ylabel("Median Value")
plt.legend()
plt.show()
```

### Visualizing Comparison of Data

**Line Chart**

Plot the median `CompTotal` for all ages from 45 to 60.


```python
## Write your code here
# Get Age and CompTotal
QUERY = """
SELECT Age, CompTotal
FROM main
WHERE Age IN ('45-54 years old', '55-64 years old')
AND CompTotal IS NOT NULL
"""

line_data = pd.read_sql_query(QUERY, conn)

# Convert compensation to numeric
line_data["CompTotal"] = pd.to_numeric(
    line_data["CompTotal"],
    errors="coerce"
)

line_data = line_data.dropna(subset=["CompTotal"])

# Calculate median compensation by age group
median_comp = (
    line_data
    .groupby("Age")["CompTotal"]
    .median()
)

# Put age groups in the correct order
age_order = [
    "45-54 years old",
    "55-64 years old"
]

median_comp = median_comp.reindex(age_order)

# Create line chart
plt.figure(figsize=(10, 6))

plt.plot(
    median_comp.index,
    median_comp.values,
    marker="o"
)

plt.title("Median Total Compensation by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Median Total Compensation")
plt.grid(True)
plt.show()
```

**Bar Chart**

Create a horizontal bar chart using the `MainBranch` column.


```python
## Write your code here
# Get MainBranch counts
QUERY = """
SELECT MainBranch, COUNT(*) AS count
FROM main
WHERE MainBranch IS NOT NULL
GROUP BY MainBranch
ORDER BY count
"""

main_branch_data = pd.read_sql_query(QUERY, conn)

# Create horizontal bar chart
plt.figure(figsize=(10, 6))

plt.barh(
    main_branch_data["MainBranch"],
    main_branch_data["count"]
)

plt.title("Respondents by Main Branch")
plt.xlabel("Number of Respondents")
plt.ylabel("Main Branch")
plt.show()
```
