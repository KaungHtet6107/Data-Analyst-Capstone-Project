### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset into a DataFrame



load the dataset using pd.read_csv()


```python
# Define the URL of the dataset
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"

# Load the dataset into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3: Identifying Duplicate Rows


**Task 1: Identify Duplicate Rows**
  1. Count the number of duplicate rows in the dataset.
  2. Display the first few duplicate rows to understand their structure.


```python
## Write your code here
# Count the number of duplicate rows
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

# Display the first few duplicate rows
duplicate_rows = df[df.duplicated(keep=False)]

print("\nFirst few duplicate rows:")
print(duplicate_rows.head())
```

### Step 4: Removing Duplicate Rows


**Task 2: Remove Duplicates**
   1. Remove duplicate rows from the dataset using the drop_duplicates() function.
2. Verify the removal by counting the number of duplicate rows after removal .


```python
## Write your code here
# Remove duplicate rows
df = df.drop_duplicates()

# Count duplicate rows after removal
duplicate_count_after = df.duplicated().sum()

print("Number of duplicate rows after removal:", duplicate_count_after)

# Display the shape of the cleaned dataset
print("Rows and columns after removing duplicates:", df.shape)
```

### Step 5: Handling Missing Values


**Task 3: Identify and Handle Missing Values**
   1. Identify missing values for all columns in the dataset.
   2. Choose a column with significant missing values (e.g., EdLevel) and impute with the most frequent value.


```python
## Write your code here
# Count missing values in all columns
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)

# Find the most frequent value in EdLevel
most_frequent_edlevel = df["EdLevel"].mode()[0]

print("\nMost frequent EdLevel:", most_frequent_edlevel)

# Fill missing EdLevel values with the most frequent value
df["EdLevel"] = df["EdLevel"].fillna(most_frequent_edlevel)

# Verify that EdLevel no longer has missing values
print("\nMissing EdLevel values after imputation:",
      df["EdLevel"].isnull().sum())
```

### Step 6: Normalizing Compensation Data


**Task 4: Normalize Compensation Data Using ConvertedCompYearly**
   1. Use the ConvertedCompYearly column for compensation analysis as the normalized annual compensation is already provided.
   2. Check for missing values in ConvertedCompYearly and handle them if necessary.


```python
## Write your code here
# Check missing values in ConvertedCompYearly
missing_compensation = df["ConvertedCompYearly"].isnull().sum()

print("Missing ConvertedCompYearly values:", missing_compensation)

# Handle missing values by filling them with the median
median_compensation = df["ConvertedCompYearly"].median()

df["ConvertedCompYearly"] = df["ConvertedCompYearly"].fillna(
    median_compensation
)

# Verify that missing values have been handled
print(
    "Missing ConvertedCompYearly values after handling:",
    df["ConvertedCompYearly"].isnull().sum()
)
```
